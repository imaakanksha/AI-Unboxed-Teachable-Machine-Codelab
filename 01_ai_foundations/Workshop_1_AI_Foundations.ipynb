## 1. How AI Actually Learns

At its core, AI does not "think" or possess conceptual comprehension. Instead, training a neural network is an iterative process of mathematical optimization called **Gradient Descent**.

### The Optimization Loop
1. **Forward Pass:** Input data is fed through layers. In a single neuron, the output is calculated as:
   $$Output = \sigma(W \cdot X + B)$$
   Where $W$ represents weights, $X$ represents inputs, $B$ represents bias, and $\sigma$ represents the activation function (e.g., ReLU or Sigmoid).
2. **Loss Function:** Compares the prediction to the true target, outputting a numerical error (loss).
   - Mean Squared Error (Regression): $L = \frac{1}{N} \sum (Y_i - \hat{Y}_i)^2$
3. **Backpropagation:** Computes how much each weight contributed to the loss using the chain rule of calculus.
4. **Optimizer Step:** Adjusts weights in the opposite direction of the gradient to reduce error:
   $$W_{new} = w_{old} - \text{learning\_rate} \times Gradient$$

### 💻 Code Exercise: Simulate Gradient Descent from Scratch

Let's write a simple Python script to train a single weight to map an input $X=3$ to a target output $Y=12$ using gradient descent. Our hypothesis is:
$$Prediction = W \cdot X$$
Our Loss function is Mean Squared Error (MSE):
$$Loss = (Prediction - Y)^2 = (W \cdot X - Y)^2$$
The gradient of Loss with respect to Weight $W$ is:
$$\frac{dLoss}{dW} = 2 \cdot X \cdot (W \cdot X - Y)$$

```python
# Initialize starting parameters
w = 1.0          # Initial weight
x = 3.0          # Input
target = 12.0    # Target label
learning_rate = 0.05

print(f"Initial Weight: {w:.2f}")
print(f"Target Output: {target:.2f}\n")

for epoch in range(1, 11):
    # 1. Forward Pass (Prediction)
    prediction = w * x
    
    # 2. Loss Calculation (Squared Error)
    loss = (prediction - target) ** 2
    
    # 3. Backpropagation (Calculate Gradient of Loss w.r.t Weight)
    gradient = 2.0 * x * (prediction - target)
    
    # 4. Weight Update
    w = w - learning_rate * gradient
    
    print(f"Epoch {epoch:02d} | Prediction: {prediction:6.2f} | Loss: {loss:6.2f} | New Weight: {w:5.2f}")
```

## 2. Why AI Feels Intelligent

Modern Large Language Models (LLMs) appear to reason because of **emergence** at scale. As models increase in parameters (weights) and training token count, they develop capabilities not explicitly programmed. 

LLMs operate via **Next-Token Prediction**. Given a context, they predict a probability distribution over the vocabulary.

```python
# Next-Token Probability Simulation
import random

context = "The capital of France is"
vocabulary = {
    "Paris": 0.95,
    "Lyon": 0.03,
    "London": 0.01,
    "Banana": 0.01
}

print(f"Context: '{context} ___'")
print("Model Vocabulary Probabilities:")
for word, prob in vocabulary.items():
    print(f"  - '{word}': {prob * 100:.1f}%")

# Sample next token according to the probability distribution
words = list(vocabulary.keys())
weights = list(vocabulary.values())
sampled_word = random.choices(words, weights=weights, k=1)[0]
print(f"\nSampled Next Token: '{sampled_word}'")
```

## 3. Hallucinations and Temperature

Hallucination occurs when a generative model outputs fluent but factually incorrect assertions. Temperature scales the logits (pre-softmax outputs) to adjust sample randomness:
$$\text{Scaled Logit} = \frac{\text{Logit}}{\text{Temperature}}$$

- **Low Temperature (<0.2):** Focuses probability on the top tokens (highly deterministic, best for facts/code).
- **High Temperature (>0.8):** Flattens the distribution (creative, higher risk of hallucinating).

```python
import numpy as np

def softmax(logits):
    exp_logits = np.exp(logits - np.max(logits)) # stable softmax
    return exp_logits / np.sum(exp_logits)

# Raw Logits for words: ['Paris', 'Lyon', 'Banana']
logits = np.array([4.0, 1.0, -1.0])

for temp in [0.2, 0.7, 1.5]:
    scaled_logits = logits / temp
    probabilities = softmax(scaled_logits)
    print(f"Temperature: {temp:.1f}")
    print(f"  - Paris: {probabilities[0]*100:5.2f}%")
    print(f"  - Lyon:  {probabilities[1]*100:5.2f}%")
    print(f"  - Banana:{probabilities[2]*100:5.2f}%")
```

## 4. Overfitting vs Underfitting

- **Underfitting (High Bias):** Model is too simple to capture trends (performs poorly on training and validation sets).
- **Overfitting (High Variance):** Model memorizes noise (performs perfectly on training set, poorly on validation set).

```python
# Visualizing Overfitting with a Polynomial Fit
import matplotlib.pyplot as plt

# Sample points (noisy line)
np.random.seed(42)
x_data = np.linspace(-3, 3, 10)
y_data = 2 * x_data + 1 + np.random.normal(0, 1.5, size=len(x_data))

# Fitting line (Simple model - Underfitting if data was complex, here it is a good fit)
p_simple = np.poly1d(np.polyfit(x_data, y_data, 1))

# Fitting a 9th degree polynomial (Memorizes all 10 points - Overfitting!)
p_complex = np.poly1d(np.polyfit(x_data, y_data, 9))

x_plot = np.linspace(-3, 3, 100)
plt.figure(figsize=(10, 5))
plt.scatter(x_data, y_data, color='red', label='Training Data')
plt.plot(x_plot, p_simple(x_plot), '--', label='Simple Linear Model')
plt.plot(x_plot, p_complex(x_plot), '-', label='9th Degree Polynomial (Overfit)')
plt.ylim(-10, 10)
plt.legend()
plt.title("Overfitting Demonstration")
plt.show()
```

## 5. AI Bias and Fairness

AI models learn bias from their training data. If a dataset is unbalanced or reflects historical human biases, the model will codify and amplify those biases. Mitigation requires deliberate data representative balancing and audit tools (like IBM AIF360).
